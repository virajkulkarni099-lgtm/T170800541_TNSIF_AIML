````python
plt.figure(figsize=(8, 5))
sns.histplot(df[target].dropna(), kde=True)
plt.title("Distribution of Parts Produced Per Hour")
plt.xlabel("Parts Per Hour")
plt.ylabel("Frequency")
plt.show()
````

````python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(14, 9))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
````

````python
df.columns = df.columns.str.strip()

print("Duplicate rows before cleaning:", df.duplicated().sum())

df = df.drop_duplicates().copy()

print("Dataset shape after removing duplicates:", df.shape)
````

````python
if "Timestamp" in df.columns:
    df["Timestamp"] = pd.to_datetime(
        df["Timestamp"],
        errors="coerce"
    )

    df["Timestamp_Hour"] = df["Timestamp"].dt.hour
    df["Timestamp_Month"] = df["Timestamp"].dt.month

    df = df.drop(columns=["Timestamp"])

print(df.columns.tolist())
````

````python
df = df.dropna(subset=[target]).copy()

print("Shape after removing missing target values:", df.shape)
print("Missing target values:", df[target].isnull().sum())
````

````python
X = df.drop(columns=[target])
y = df[target]

print("Input features shape:", X.shape)
print("Target shape:", y.shape)

display(X.head())
display(y.head())
````

````python
numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Numerical features:", numeric_features)
print("\nCategorical features:", categorical_features)
````

````python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

print("Preprocessing pipeline created successfully!")
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training input:", X_train.shape)
print("Testing input:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
````

````python
model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
````

````python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual_Parts_Per_Hour": y_test.values,
    "Predicted_Parts_Per_Hour": y_pred
})

results["Prediction_Error"] = (
    results["Actual_Parts_Per_Hour"]
    - results["Predicted_Parts_Per_Hour"]
)

display(results.head(10))
````

````python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")
print("------------------------")
print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R²   : {r2:.4f}")
print(f"R² percentage: {r2 * 100:.2f}%")
````

### Part 2: Advanced Preprocessing with Feature Scaling & Regularized Models

In this section, we will optimize our preprocessing pipeline by adding `StandardScaler` to the numerical features. This ensures that features with larger numeric ranges do not dominate the regression coefficients. We will also train a **Ridge Regression** model to introduce L2 regularization and compare its performance with our basic Linear Regression model.

````python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

# 1. Update the numerical transformer to include Feature Scaling (Standardization)
numeric_transformer_scaled = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# 2. Re-create preprocessor
preprocessor_scaled = ColumnTransformer(transformers=[
    ("num", numeric_transformer_scaled, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# 3. Define the new Pipeline with Ridge Regression
ridge_model = Pipeline(steps=[
    ("preprocessor", preprocessor_scaled),
    ("regressor", Ridge(alpha=1.0))
])

# 4. Train the regularized model
ridge_model.fit(X_train, y_train)
print("Ridge Regression model with standardized features trained successfully!")
````

````python
# Evaluate the Ridge model
y_pred_ridge = ridge_model.predict(X_test)

mae_r = mean_absolute_error(y_test, y_pred_ridge)
mse_r = mean_squared_error(y_test, y_pred_ridge)
rmse_r = np.sqrt(mse_r)
r2_r = r2_score(y_test, y_pred_ridge)

# Compare with previous Linear Regression
comparison_updated = pd.DataFrame({
    "Model": ["Mean Baseline", "Linear Regression (No Scaling)", "Ridge Regression (With Scaling)"],
    "MAE": [baseline_mae, mae, mae_r],
    "RMSE": [baseline_rmse, rmse, rmse_r],
    "R2": [baseline_r2, r2, r2_r]
})

display(comparison_updated)
````

### Part 3: Deploying the Model with FastAPI

To fulfill the Capstone deployment requirement, we will write a production-ready `app.py` script containing a FastAPI application. We will use the saved model pipeline to serve real-time predictions. Then, we will start the FastAPI server in the background and test it with a sample API request.

````python
# 1. Install required web serving libraries
!pip install fastapi uvicorn nest-asyncio -q

import nest_asyncio
nest_asyncio.apply()
print("Libraries installed and async environment configured!")
````

````python
%%writefile app.py
import joblib
import pandas as pd
from fastapi import FastAPI
from pydantic import BaseModel
from typing import Optional

# Initialize FastAPI app
app = FastAPI(title="Manufacturing Equipment Output Predictor API")

# Load the trained model pipeline
model = joblib.load("manufacturing_output_model.pkl")

# Define the input schema using Pydantic
class ProductionParameters(BaseModel):
    Injection_Temperature: float
    Injection_Pressure: float
    Cycle_Time: float
    Cooling_Time: float
    Material_Viscosity: Optional[float] = None
    Ambient_Temperature: Optional[float] = None
    Machine_Age: float
    Operator_Experience: Optional[float] = None
    Maintenance_Hours: float
    Shift: str
    Machine_Type: str
    Material_Grade: str
    Day_of_Week: str
    Temperature_Pressure_Ratio: float
    Total_Cycle_Time: float
    Efficiency_Score: float
    Machine_Utilization: float
    Timestamp_Hour: int
    Timestamp_Month: int

@app.get("/")
def home():
    return {"message": "Manufacturing Output Prediction API is online! Use the /predict endpoint for inference."}

@app.post("/predict")
def predict(data: ProductionParameters):
    # Convert incoming request data to a Pandas DataFrame
    input_dict = data.dict()
    input_df = pd.DataFrame([input_dict])

    # Generate prediction using the pipeline
    prediction = model.predict(input_df)[0]

    return {
        "status": "success",
        "predicted_parts_per_hour": round(prediction, 2)
    }
````

````python
import subprocess
import time
import requests

# Start FastAPI app in the background on port 8000
proc = subprocess.Popen(["uvicorn", "app:app", "--host", "127.0.0.1", "--port", "8000"])

# Give the server a few seconds to start up
time.sleep(3)

print("FastAPI server is running in the background!")
````

````python
# Let's construct a sample payload based on our test dataset to verify the API endpoint
sample_payload = {
    "Injection_Temperature": 221.5,
    "Injection_Pressure": 123.7,
    "Cycle_Time": 40.9,
    "Cooling_Time": 11.0,
    "Material_Viscosity": 144.9,
    "Ambient_Temperature": 22.1,
    "Machine_Age": 5.0,
    "Operator_Experience": 10.0,
    "Maintenance_Hours": 15.0,
    "Shift": "Day",
    "Machine_Type": "Type_A",
    "Material_Grade": "Standard",
    "Day_of_Week": "Monday",
    "Temperature_Pressure_Ratio": 1.79,
    "Total_Cycle_Time": 51.9,
    "Efficiency_Score": 0.85,
    "Machine_Utilization": 0.78,
    "Timestamp_Hour": 14,
    "Timestamp_Month": 5
}

# Query our local API server
url = "http://127.0.0.1:8000/predict"
try:
    response = requests.post(url, json=sample_payload)
    print("API Status Code:", response.status_code)
    print("API Response:", response.json())
finally:
    # Clean up the background server process
    proc.terminate()
    print("Server successfully stopped.")
````

### Local Version Fix: Retrain and Save Model File Locally

If you get pickling errors in VS Code due to version differences, run this script on your local machine to generate a perfect `manufacturing_output_model.pkl` matching your exact local Python and scikit-learn environments.

````python
%%writefile retrain_model.py
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
import joblib

# 1. Load the original raw dataset
try:
    df = pd.read_csv("manufacturing_dataset_1000_samples.csv")
except FileNotFoundError:
    print("Error: Place 'manufacturing_dataset_1000_samples.csv' in the same folder as this script!")
    exit(1)

# 2. Re-apply cleaning & preprocessing transformations
df.columns = df.columns.str.strip()

if "Timestamp" in df.columns:
    df["Timestamp"] = pd.to_datetime(df["Timestamp"], errors="coerce")
    df["Timestamp_Hour"] = df["Timestamp"].dt.hour
    df["Timestamp_Month"] = df["Timestamp"].dt.month
    df = df.drop(columns=["Timestamp"])

target = "Parts_Per_Hour"
X = df.drop(columns=[target])
y = df[target]

numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

# 3. Rebuild the pipeline
numeric_transformer_scaled = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_scaled = ColumnTransformer(transformers=[
    ("num", numeric_transformer_scaled, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

ridge_model = Pipeline(steps=[
    ("preprocessor", preprocessor_scaled),
    ("regressor", Ridge(alpha=1.0))
])

# 4. Train the model locally
print("Training Ridge Regression pipeline on your local environment...")
ridge_model.fit(X, y)

# 5. Export clean model
joblib.dump(ridge_model, "manufacturing_output_model.pkl")
print("Success! A new, compatible 'manufacturing_output_model.pkl' has been generated locally!")
````
