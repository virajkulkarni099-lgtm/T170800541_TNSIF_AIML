```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["slappy", "boss", "ben", "badass", "aleinx", None],
    "Age": [21,22,23,34,25,np.nan],
    "City": ["pune", "mumbai", "delhi","mumbai", "pune","pune"]

}
print(data)
df = pd.DataFrame(data)
df
```

```python
df.isnull()
df.isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Name"] = df["Name"].fillna(df["Name"].mode()[0])

df
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "Salary": [1000,2000,3000,4000,5000,6000],
    "Age": [21,22,23,34,25,np.nan],
    "City": ["pune", "mumbai", "delhi", "mumbai", "pune", "pune"]

})
df
```

```python
import pandas as pd

df = pd.read_csv("rainfall_districtwise_daily_imd.csv")

print(df.head())
print(df.tail())
```

```python
print("Rows and Columns:", df.shape)

print("\nData Types:")
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum()[df.isnull().sum() > 0])
```

filling missing values

```python
num_cols = df.select_dtypes(include="number").columns
cat_cols = df.select_dtypes(include="object").columns
```

```python
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```
