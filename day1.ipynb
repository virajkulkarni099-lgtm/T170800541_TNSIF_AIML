```python

class Dog:
    species = "Canine"

    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Buddy", 3)
dog2 = Dog("Miky", 7)

print(dog1.name)
print(dog1.age)
print(dog1.species)

print(dog2.name)
print(dog2.age)
```

```python

class Details:
    def __init__(self, name, age, height, weight):
        self.name = name
        self.age = age
        self.height = height
        self.weight = weight

        print(
            f"The name of the student is {self.name} "
            f"and age is {self.age}"
        )
        print(
            f"The height is {self.height} "
            f"and weight is {self.weight}"
        )

    def student(self):
        print(
            f"The name of the student is {self.name} "
            f"and age is {self.age}"
        )
        print(
            f"The height is {self.height} "
            f"and weight is {self.weight}"
        )

    def person(self, student_id, roll_no):
        self.student_id = student_id
        self.roll_no = roll_no

        print(
            f"The ID is {self.student_id}, "
            f"roll number is {self.roll_no}, "
            f"and name is {self.name}"
        )


a = Details("Viraj", 20, 175.67, 60)

a.student()
a.person(1001, 10)
```
