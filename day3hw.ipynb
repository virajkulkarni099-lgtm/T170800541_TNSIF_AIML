```python
#Reverse a list without using built-in functions.

num = [2,3,2,3,2,4,32,2,4,5,4,5,63,2,42,3]
rev = []

for i in range (len(num) -1,-1,-1):
  rev.append(num[i])
print(rev)
```

```python
#Perform union, intersection of two sets.

set1 = {23,24,2,4,1,1,4,2,4,}
set2 ={3,2,4,2,2,4,1,3,2,2,2,4,4,5,67}
union = set1.union(set2)
print("union:",union)
intersection = set1.intersection(set2)
print("intersection:",intersection)
```

```python
#Check if one set is subset of another
set1 = {23,24,2,4,1,1,4,2,4,}
set2 ={3,2,4,2,2,4,1,3,2,2,2,4,4,5,67}
subset = set1.issubset(set2)
print("subset:",subset)
```

```python
class Rectangle:
  def __init__(self, width, length):
    self.length = length
    self.width = width
  def area(self):
    return self.length * self.width
  def perimeter(self):
    return 2*(self.length + self.width)
obj1 = Rectangle(12,6)
print(obj1.area())
print(obj1.perimeter())
```
