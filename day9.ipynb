```python

# ============================================================
# DAY 9: MACHINE LEARNING INTRODUCTION — COMPLETE PRACTICAL
# Google Colab: Run this entire cell from top to bottom
# ============================================================

# STEP 1: IMPORT LIBRARIES
import os
import glob
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display
from google.colab import files

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix,
    mean_squared_error, mean_absolute_error, r2_score,
    silhouette_score
)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")


# STEP 2: INTRODUCTION TO MACHINE LEARNING
print("\n========== TYPES OF MACHINE LEARNING ==========")

ml_types = pd.DataFrame({
    "Type": [
        "Supervised Learning",
        "Unsupervised Learning",
        "Semi-Supervised Learning",
        "Reinforcement Learning"
    ],
    "Data": [
        "Labeled data",
        "Unlabeled data",
        "Small labeled + large unlabeled",
        "Interaction and rewards"
    ],
    "Goal": [
        "Predict outputs",
        "Discover patterns",
        "Learn with limited labels",
        "Choose rewarding actions"
    ],
    "Example": [
        "Spam detection",
        "Customer segmentation",
        "Image classification",
        "Game-playing AI"
    ]
})
display(ml_types)

print("""
Supervised algorithms: Linear Regression, Logistic Regression,
Decision Tree, Random Forest, SVM, KNN, Naive Bayes.

Unsupervised algorithms: K-Means, Hierarchical Clustering,
DBSCAN, PCA.

Semi-supervised techniques: Self-Training, Label Propagation.

Reinforcement learning algorithms: Q-Learning, SARSA, DQN.

Deep Learning uses multilayer neural networks.
NLP focuses on processing human language.
""")


# STEP 3: LOAD THE RAINFALL DATASET
print("\n========== LOADING DATASET ==========")

possible_paths = [
    "/content/rainfall_dataset/rainfall.csv",
    "/content/rainfall.csv"
]

csv_path = next(
    (path for path in possible_paths if os.path.exists(path)),
    None
)

# Search for another CSV in the Colab working directory
if csv_path is None:
    csv_files = glob.glob("/content/**/*.csv", recursive=True)
    if csv_files:
        csv_path = csv_files[0]

if csv_path is None:
    print("Upload your rainfall CSV file.")
    uploaded = files.upload()
    csv_files = [
        name for name in uploaded
        if name.lower().endswith(".csv")
    ]

    if not csv_files:
        raise ValueError("No CSV file uploaded. Please upload a CSV dataset.")

    csv_path = csv_files[0]

df = pd.read_csv(csv_path)

# Standardize column names
df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Dataset loaded from:", csv_path)
print("Original shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())


# STEP 4: DATA PREPROCESSING
print("\n========== DATA PREPROCESSING ==========")

# Remove duplicate rows
duplicate_count = df.duplicated().sum()
df = df.drop_duplicates().copy()
print("Duplicate rows removed:", duplicate_count)

# Convert date column if available
if "date" in df.columns:
    df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Convert expected numeric columns to numeric format
numeric_candidates = [
    "rainfall", "temperature", "humidity", "wind_speed"
]

for col in numeric_candidates:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove rows where the classification target is missing
if "weather_condition" in df.columns:
    df["weather_condition"] = (
        df["weather_condition"].astype("string").str.strip()
    )
    df["weather_condition"] = df["weather_condition"].replace(
        "", pd.NA
    )
    df = df.dropna(subset=["weather_condition"]).copy()

# Remove rows where all available numeric features are missing
available_numeric = [
    col for col in numeric_candidates if col in df.columns
]

if available_numeric:
    df = df.dropna(
        subset=available_numeric,
        how="all"
    ).copy()

if df.empty:
    raise ValueError(
        "No usable rows remain after cleaning. Check your CSV file."
    )

print("Cleaned shape:", df.shape)
print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").transpose())


# STEP 5: EXPLORATORY DATA ANALYSIS (EDA)
print("\n========== EXPLORATORY DATA ANALYSIS ==========")

print("First five rows:")
display(df.head())

print("Last five rows:")
display(df.tail())

print("Dataset shape:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn data types:")
display(df.dtypes.to_frame("Data Type"))

print("\nUnique values per column:")
display(df.nunique().to_frame("Unique Values"))

print("\nMissing-value count:")
display(df.isnull().sum().to_frame("Missing Values"))

# Numeric distributions
numeric_df = df.select_dtypes(include=np.number)

if not numeric_df.empty:
    numeric_df.hist(figsize=(12, 8), bins=15)
    plt.suptitle("Distribution of Numeric Features")
    plt.tight_layout()
    plt.show()

    # Correlation heatmap
    if numeric_df.shape[1] >= 2:
        plt.figure(figsize=(8, 6))
        sns.heatmap(
            numeric_df.corr(),
            annot=True,
            cmap="coolwarm",
            fmt=".2f"
        )
        plt.title("Correlation Heatmap")
        plt.show()

# Target distribution
if "weather_condition" in df.columns:
    print("\nWeather condition counts:")
    display(df["weather_condition"].value_counts().to_frame("Count"))

    sns.countplot(
        data=df,
        x="weather_condition",
        hue="weather_condition",
        legend=False
    )
    plt.title("Weather Condition Distribution")
    plt.xticks(rotation=30)
    plt.show()

# Boxplots for numeric features
if not numeric_df.empty:
    for col in numeric_df.columns:
        plt.figure(figsize=(7, 3))
        sns.boxplot(x=df[col])
        plt.title(f"Boxplot: {col}")
        plt.show()

# Average rainfall by weather condition
if (
    "rainfall" in df.columns
    and "weather_condition" in df.columns
):
    print("\nAverage rainfall by weather condition:")
    display(
        df.groupby("weather_condition")["rainfall"]
        .mean()
        .to_frame("Average Rainfall")
    )


# STEP 6: SUPERVISED LEARNING — CLASSIFICATION
print("\n========== CLASSIFICATION ==========")

feature_cols = [
    col for col in [
        "rainfall", "temperature", "humidity", "wind_speed"
    ]
    if col in df.columns
]

if "weather_condition" in df.columns and len(feature_cols) >= 1:
    X = df[feature_cols]
    y = df["weather_condition"].astype(str)

    # Remove classes with fewer than 2 examples because stratified
    # train/test splitting requires enough examples in each class.
    class_counts = y.value_counts()
    valid_classes = class_counts[class_counts >= 2].index

    classification_data = df[y.isin(valid_classes)].copy()
    X = classification_data[feature_cols]
    y = classification_data["weather_condition"].astype(str)

    if y.nunique() >= 2 and y.value_counts().min() >= 2:
        X_train, X_test, y_train, y_test = train_test_split(
            X,
            y,
            test_size=0.20,
            random_state=42,
            stratify=y
        )

        # Pipeline: fill missing values -> scale -> train classifier
        logistic_model = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(max_iter=1000))
        ])

        logistic_model.fit(X_train, y_train)
        y_pred = logistic_model.predict(X_test)

        print("Features used:", feature_cols)
        print("Training rows:", len(X_train))
        print("Testing rows:", len(X_test))
        print("\nAccuracy:", round(accuracy_score(y_test, y_pred), 4))
        print(
            "Precision (weighted):",
            round(
                precision_score(
                    y_test, y_pred,
                    average="weighted",
                    zero_division=0
                ), 4
            )
        )
        print(
            "Recall (weighted):",
            round(
                recall_score(
                    y_test, y_pred,
                    average="weighted",
                    zero_division=0
                ), 4
            )
        )
        print(
            "F1 Score (weighted):",
            round(
                f1_score(
                    y_test, y_pred,
                    average="weighted",
                    zero_division=0
                ), 4
            )
        )

        print("\nClassification report:")
        print(classification_report(
            y_test, y_pred, zero_division=0
        ))

        # Confusion matrix
        labels = sorted(y.unique())
        cm = confusion_matrix(y_test, y_pred, labels=labels)

        plt.figure(figsize=(6, 4))
        sns.heatmap(
            cm,
            annot=True,
            fmt="d",
            cmap="Blues",
            xticklabels=labels,
            yticklabels=labels
        )
        plt.xlabel("Predicted")
        plt.ylabel("Actual")
        plt.title("Confusion Matrix — Logistic Regression")
        plt.show()

        # Compare with Decision Tree
        tree_model = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", DecisionTreeClassifier(
                max_depth=3,
                random_state=42
            ))
        ])

        tree_model.fit(X_train, y_train)
        tree_pred = tree_model.predict(X_test)

        comparison = pd.DataFrame({
            "Model": ["Logistic Regression", "Decision Tree"],
            "Accuracy": [
                accuracy_score(y_test, y_pred),
                accuracy_score(y_test, tree_pred)
            ],
            "F1 Weighted": [
                f1_score(
                    y_test, y_pred,
                    average="weighted",
                    zero_division=0
                ),
                f1_score(
                    y_test, tree_pred,
                    average="weighted",
                    zero_division=0
                )
            ]
        })

        print("\nModel comparison:")
        display(comparison)

        # Cross-validation, using 3 folds when every class has
        # at least 3 samples; otherwise use 2 folds.
        min_class_count = y.value_counts().min()
        folds = min(3, int(min_class_count))

        cv_scores = cross_val_score(
            logistic_model,
            X,
            y,
            cv=folds,
            scoring="accuracy"
        )

        print(f"\n{folds}-Fold CV accuracy scores:", cv_scores)
        print("Mean CV accuracy:", round(cv_scores.mean(), 4))

    else:
        print(
            "Classification skipped: the dataset needs at least "
            "two classes with enough examples."
        )
        logistic_model = None

else:
    print(
        "Classification skipped: expected weather_condition "
        "and at least one numeric feature."
    )
    logistic_model = None


# STEP 7: SUPERVISED LEARNING — REGRESSION
print("\n========== REGRESSION ==========")

regression_features = [
    col for col in ["temperature", "humidity", "wind_speed"]
    if col in df.columns
]

if "rainfall" in df.columns and len(regression_features) >= 1:
    regression_data = df[
        regression_features + ["rainfall"]
    ].copy()

    regression_data["rainfall"] = pd.to_numeric(
        regression_data["rainfall"], errors="coerce"
    )
    regression_data = regression_data.dropna(
        subset=["rainfall"]
    )

    if len(regression_data) >= 5:
        X_reg = regression_data[regression_features]
        y_reg = regression_data["rainfall"]

        X_reg_train, X_reg_test, y_reg_train, y_reg_test = (
            train_test_split(
                X_reg, y_reg,
                test_size=0.20,
                random_state=42
            )
        )

        regression_model = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", LinearRegression())
        ])

        regression_model.fit(X_reg_train, y_reg_train)
        rainfall_pred = regression_model.predict(X_reg_test)

        mse = mean_squared_error(y_reg_test, rainfall_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_reg_test, rainfall_pred)
        r2 = r2_score(y_reg_test, rainfall_pred)

        print("Regression features:", regression_features)
        print("MSE:", round(mse, 4))
        print("RMSE:", round(rmse, 4))
        print("MAE:", round(mae, 4))
        print("R² Score:", round(r2, 4))

        regression_results = pd.DataFrame({
            "Actual Rainfall": y_reg_test.to_numpy(),
            "Predicted Rainfall": rainfall_pred
        })

        print("\nActual vs predicted rainfall:")
        display(regression_results)

        plt.figure(figsize=(7, 5))
        plt.scatter(y_reg_test, rainfall_pred)
        plt.xlabel("Actual Rainfall")
        plt.ylabel("Predicted Rainfall")
        plt.title("Actual vs Predicted Rainfall")
        plt.show()

    else:
        print("Regression skipped: not enough valid rows.")
        regression_model = None
else:
    print("Regression skipped: required columns are missing.")
    regression_model = None


# STEP 8: UNSUPERVISED LEARNING — K-MEANS CLUSTERING
print("\n========== K-MEANS CLUSTERING ==========")

cluster_features = [
    col for col in [
        "rainfall", "temperature", "humidity", "wind_speed"
    ]
    if col in df.columns
]

cluster_data = df[cluster_features].copy()
cluster_data = cluster_data.dropna(how="all")

if len(cluster_data) >= 3 and len(cluster_features) >= 2:
    cluster_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    X_cluster = cluster_pipeline.fit_transform(cluster_data)

    # Three clusters, provided there are enough rows
    kmeans = KMeans(
        n_clusters=3,
        random_state=42,
        n_init=10
    )

    cluster_labels = kmeans.fit_predict(X_cluster)

    cluster_results = cluster_data.copy()
    cluster_results["cluster"] = cluster_labels

    print("Cluster assignments:")
    display(cluster_results.head(10))

    print("\nNumber of rows per cluster:")
    display(
        cluster_results["cluster"]
        .value_counts()
        .sort_index()
        .to_frame("Rows")
    )

    if len(np.unique(cluster_labels)) > 1:
        score = silhouette_score(X_cluster, cluster_labels)
        print("Silhouette score:", round(score, 4))
else:
    print("Clustering skipped: not enough numeric data.")


# STEP 9: SAVE THE TRAINED CLASSIFICATION MODEL
print("\n========== SAVE MODEL ==========")

if logistic_model is not None:
    model_filename = "/content/model.pkl"

    with open(model_filename, "wb") as model_file:
        pickle.dump(logistic_model, model_file)

    print("Model saved successfully:", model_filename)

    # Load the saved model again
    with open(model_filename, "rb") as model_file:
        loaded_model = pickle.load(model_file)

    print("Saved model loaded successfully!")

    # Test the loaded model using one real dataset row
    sample_input = X.iloc[[0]]
    sample_prediction = loaded_model.predict(sample_input)

    print("\nSample input:")
    display(sample_input)

    print("Predicted weather condition:", sample_prediction[0])

    # Compare with actual value if available
    print("Actual weather condition:", y.iloc[0])

else:
    print(
        "No classification model was trained, so model.pkl "
        "was not created."
    )


# STEP 10: DOWNLOAD THE MODEL
print("\n========== DOWNLOAD ==========")

if logistic_model is not None and os.path.exists("/content/model.pkl"):
    print("Downloading model.pkl...")
    files.download("/content/model.pkl")
else:
    print("No saved classification model available to download.")



```

```python

# ============================================================
# DAY 9 - ADVANCED TOPICS AND REMAINING ML PIPELINE
# Run after the previous Day 9 cell
# ============================================================

import os
import json
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.semi_supervised import SelfTrainingClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.metrics import f1_score
from scipy.stats import ks_2samp

warnings.filterwarnings("ignore")
np.random.seed(42)

print("Advanced ML libraries imported successfully!")


# ------------------------------------------------------------
# STEP 1: HYPERPARAMETER TUNING
# GridSearchCV finds the best parameters using cross-validation
# ------------------------------------------------------------

print("\n===== 1. HYPERPARAMETER TUNING =====")

feature_cols = [
    c for c in ["rainfall", "temperature", "humidity", "wind_speed"]
    if c in df.columns
]

if (
    "weather_condition" in df.columns
    and len(feature_cols) > 0
    and df["weather_condition"].nunique() >= 2
):
    tuning_df = df[feature_cols + ["weather_condition"]].copy()
    tuning_df["weather_condition"] = (
        tuning_df["weather_condition"].astype(str).str.strip()
    )
    tuning_df = tuning_df.replace([np.inf, -np.inf], np.nan)
    tuning_df = tuning_df.dropna(subset=["weather_condition"])

    X_tune = tuning_df[feature_cols]
    y_tune = tuning_df["weather_condition"]

    counts = y_tune.value_counts()
    valid_classes = counts[counts >= 2].index
    X_tune = X_tune[y_tune.isin(valid_classes)]
    y_tune = y_tune[y_tune.isin(valid_classes)]

    if y_tune.nunique() >= 2:
        folds = min(3, int(y_tune.value_counts().min()))

        tune_pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(max_iter=1000))
        ])

        parameters = {
            "model__C": [0.1, 1.0, 10.0],
            "model__solver": ["lbfgs"]
        }

        grid = GridSearchCV(
            tune_pipeline,
            parameters,
            cv=folds,
            scoring="accuracy"
        )

        grid.fit(X_tune, y_tune)

        print("Best parameters:", grid.best_params_)
        print("Best cross-validation accuracy:", grid.best_score_)

        best_model = grid.best_estimator_
        with open("/content/best_model.pkl", "wb") as f:
            pickle.dump(best_model, f)

        print("Tuned model saved as best_model.pkl")
    else:
        best_model = None
        print("Tuning skipped: insufficient examples per class.")
else:
    best_model = None
    print("Tuning skipped: required classification columns are missing.")


# ------------------------------------------------------------
# STEP 2: SEMI-SUPERVISED LEARNING
# Hide some labels and let the model estimate them
# ------------------------------------------------------------

print("\n===== 2. SEMI-SUPERVISED LEARNING =====")

if (
    "weather_condition" in df.columns
    and len(feature_cols) > 0
    and df["weather_condition"].nunique() >= 2
):
    semi_df = df[feature_cols + ["weather_condition"]].copy()
    semi_df["weather_condition"] = (
        semi_df["weather_condition"].astype(str).str.strip()
    )
    semi_df = semi_df.dropna(subset=["weather_condition"])
    semi_df = semi_df.replace([np.inf, -np.inf], np.nan)

    # Only retain classes with at least two rows
    counts = semi_df["weather_condition"].value_counts()
    semi_df = semi_df[
        semi_df["weather_condition"].isin(
            counts[counts >= 2].index
        )
    ].copy()

    if semi_df["weather_condition"].nunique() >= 2:
        X_semi = semi_df[feature_cols].reset_index(drop=True)
        y_semi = semi_df["weather_condition"].reset_index(drop=True)

        rng = np.random.default_rng(42)
        hidden_count = max(1, int(len(y_semi) * 0.30))
        hidden_indices = rng.choice(
            len(y_semi), size=hidden_count, replace=False
        )

        y_partially_labeled = y_semi.to_numpy(dtype=object).copy()
        y_partially_labeled[hidden_indices] = -1

        semi_pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", SelfTrainingClassifier(
                estimator=LogisticRegression(max_iter=1000),
                max_iter=10
            ))
        ])

        semi_pipeline.fit(X_semi, y_partially_labeled)
        semi_predictions = semi_pipeline.predict(X_semi)

        print("Total rows:", len(X_semi))
        print("Rows with hidden labels:", hidden_count)
        print("Rows with labels initially available:",
              len(X_semi) - hidden_count)
        print("Example predictions:", semi_predictions[:10])
        print(
            "This is a demonstration: hidden labels are simulated, "
            "not genuine missing annotations."
        )
    else:
        print("Semi-supervised learning skipped: not enough classes.")
else:
    print("Semi-supervised learning skipped: dataset requirements not met.")


# ------------------------------------------------------------
# STEP 3: REINFORCEMENT LEARNING
# Small Q-learning example: an agent learns to reach a goal
# ------------------------------------------------------------

print("\n===== 3. REINFORCEMENT LEARNING: Q-LEARNING =====")

# States: 0, 1, 2, 3, 4
# Actions: 0 = move left, 1 = move right
# Goal state: 4

states = 5
actions = 2
goal_state = 4

Q = np.zeros((states, actions))

learning_rate = 0.1
discount_factor = 0.9
epsilon = 0.2
episodes = 500

for episode in range(episodes):
    state = 0

    for step in range(30):
        if np.random.random() < epsilon:
            action = np.random.randint(actions)
        else:
            action = np.argmax(Q[state])

        if action == 0:
            next_state = max(0, state - 1)
        else:
            next_state = min(goal_state, state + 1)

        reward = 10 if next_state == goal_state else -1
        done = next_state == goal_state

        Q[state, action] += learning_rate * (
            reward
            + discount_factor * np.max(Q[next_state]) * (not done)
            - Q[state, action]
        )

        state = next_state

        if done:
            break

print("Learned Q-table:")
print(np.round(Q, 2))

state = 0
path = [state]

for _ in range(10):
    action = np.argmax(Q[state])
    state = max(0, state - 1) if action == 0 else min(4, state + 1)
    path.append(state)

    if state == goal_state:
        break

print("Learned route:", path)
print("Goal reached:", state == goal_state)


# ------------------------------------------------------------
# STEP 4: DEEP LEARNING INTRODUCTION
# MLPClassifier is a neural network implementation in scikit-learn
# ------------------------------------------------------------

print("\n===== 4. DEEP LEARNING: NEURAL NETWORK =====")

if (
    "weather_condition" in df.columns
    and len(feature_cols) > 0
):
    nn_df = df[feature_cols + ["weather_condition"]].copy()
    nn_df["weather_condition"] = (
        nn_df["weather_condition"].astype(str).str.strip()
    )
    nn_df = nn_df.dropna(subset=["weather_condition"])
    nn_df = nn_df.replace([np.inf, -np.inf], np.nan)

    counts = nn_df["weather_condition"].value_counts()
    nn_df = nn_df[
        nn_df["weather_condition"].isin(counts[counts >= 2].index)
    ]

    X_nn = nn_df[feature_cols]
    y_nn = nn_df["weather_condition"]

    if y_nn.nunique() >= 2:
        X_train_nn, X_test_nn, y_train_nn, y_test_nn = (
            train_test_split(
                X_nn, y_nn,
                test_size=0.2,
                random_state=42,
                stratify=y_nn
            )
        )

        neural_network = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", MLPClassifier(
                hidden_layer_sizes=(16, 8),
                max_iter=500,
                random_state=42,
                early_stopping=False
            ))
        ])

        neural_network.fit(X_train_nn, y_train_nn)
        nn_predictions = neural_network.predict(X_test_nn)

        print("Neural network accuracy:",
              accuracy_score(y_test_nn, nn_predictions))
        print(classification_report(
            y_test_nn, nn_predictions, zero_division=0
        ))

        with open("/content/neural_network.pkl", "wb") as f:
            pickle.dump(neural_network, f)

        print("Neural network saved as neural_network.pkl")
else:
    print("Neural network skipped: required data is missing.")

print("""
Other deep-learning architectures:
- CNN: commonly used for images
- RNN/LSTM/GRU: designed for sequential data
- Transformers: widely used in modern language models
This example trains a small multilayer perceptron, not a CNN or Transformer.
""")


# ------------------------------------------------------------
# STEP 5: NLP — TEXT CLASSIFICATION
# TF-IDF converts text into numeric features
# ------------------------------------------------------------

print("\n===== 5. NLP: TEXT CLASSIFICATION =====")

texts = [
    "I love this product",
    "This is a wonderful experience",
    "The service is excellent",
    "I am very happy",
    "This is fantastic",
    "I hate this product",
    "The service is terrible",
    "This is a bad experience",
    "I am very disappointed",
    "This is awful",
    "The product is good",
    "The product is poor"
]

labels = [
    "positive", "positive", "positive", "positive", "positive",
    "negative", "negative", "negative", "negative", "negative",
    "positive", "negative"
]

text_vectorizer = TfidfVectorizer()
X_text = text_vectorizer.fit_transform(texts)

text_model = LogisticRegression(max_iter=1000)
text_model.fit(X_text, labels)

new_texts = [
    "This product is excellent",
    "I hate this terrible service"
]

new_vectors = text_vectorizer.transform(new_texts)
text_predictions = text_model.predict(new_vectors)

for text, prediction in zip(new_texts, text_predictions):
    print("Text:", text)
    print("Predicted sentiment:", prediction)
    print()

with open("/content/nlp_model.pkl", "wb") as f:
    pickle.dump(
        {
            "vectorizer": text_vectorizer,
            "model": text_model
        },
        f
    )

print("NLP model saved as nlp_model.pkl")
print("Note: this is a tiny teaching dataset, not a production sentiment model.")


# ------------------------------------------------------------
# STEP 6: MONITORING MODEL PERFORMANCE
# Compare predictions with known outcomes
# ------------------------------------------------------------

print("\n===== 6. MODEL MONITORING =====")

if best_model is not None:
    X_monitor_train, X_monitor_test, y_monitor_train, y_monitor_test = (
        train_test_split(
            X_tune,
            y_tune,
            test_size=0.2,
            random_state=7,
            stratify=y_tune
        )
    )

    monitoring_model = best_model
    monitoring_model.fit(X_monitor_train, y_monitor_train)
    monitor_pred = monitoring_model.predict(X_monitor_test)

    monitoring_accuracy = accuracy_score(
        y_monitor_test, monitor_pred
    )
    monitoring_f1 = f1_score(
        y_monitor_test,
        monitor_pred,
        average="weighted",
        zero_division=0
    )

    monitoring_report = {
        "accuracy": float(monitoring_accuracy),
        "weighted_f1": float(monitoring_f1),
        "test_rows": int(len(X_monitor_test))
    }

    with open("/content/monitoring_report.json", "w") as f:
        json.dump(monitoring_report, f, indent=4)

    print("Monitoring metrics:")
    print(json.dumps(monitoring_report, indent=4))
    print("Report saved as monitoring_report.json")
else:
    print("Monitoring skipped because tuning model is unavailable.")


# ------------------------------------------------------------
# STEP 7: DATA DRIFT DETECTION
# Compare feature distributions with the training distribution
# ------------------------------------------------------------

print("\n===== 7. DATA DRIFT DETECTION =====")

if len(feature_cols) > 0:
    drift_df = df[feature_cols].copy()
    drift_df = drift_df.replace([np.inf, -np.inf], np.nan)

    drift_baseline = drift_df.dropna(how="all").copy()

    if len(drift_baseline) >= 10:
        baseline = drift_baseline.sample(
            frac=0.5, random_state=42
        )
        current = drift_baseline.drop(baseline.index)

        drift_results = []

        for col in feature_cols:
            old_values = baseline[col].dropna()
            new_values = current[col].dropna()

            if len(old_values) >= 2 and len(new_values) >= 2:
                statistic, p_value = ks_2samp(
                    old_values, new_values
                )

                drift_results.append({
                    "feature": col,
                    "KS statistic": statistic,
                    "p-value": p_value,
                    "possible_drift": bool(p_value < 0.05)
                })

        drift_report = pd.DataFrame(drift_results)
        display(drift_report)

        print(
            "A small p-value can indicate a distribution difference. "
            "This split is only a demonstration; it is not a real "
            "historical-versus-live drift assessment."
        )
    else:
        print("Drift detection skipped: not enough rows.")
else:
    print("Drift detection skipped: no numeric features.")


# ------------------------------------------------------------
# STEP 8: RETRAINING PIPELINE
# Refit the model using the available cleaned dataset
# ------------------------------------------------------------

print("\n===== 8. MODEL RETRAINING =====")

if best_model is not None:
    # This demonstrates refitting on available data.
    # In production, use newly collected and validated labeled data.
    retrained_model = best_model
    retrained_model.fit(X_tune, y_tune)

    with open("/content/retrained_model.pkl", "wb") as f:
        pickle.dump(retrained_model, f)

    print("Retraining demonstration completed.")
    print("Saved as retrained_model.pkl")
    print(
        "Important: refitting on the same dataset does not prove "
        "improved performance. Evaluate on a separate test set."
    )
else:
    print("Retraining skipped because no suitable model was trained.")


# ------------------------------------------------------------
# STEP 9: CREATE A FASTAPI MODEL API
# Creates an app.py file for later running/deployment
# ------------------------------------------------------------

print("\n===== 9. CREATE FASTAPI API FILE =====")

if best_model is not None:
    api_code = '''
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
import pickle
import pandas as pd

app = FastAPI(title="Rainfall ML Prediction API")

with open("/content/best_model.pkl", "rb") as f:
    model = pickle.load(f)

FEATURES = ''' + repr(feature_cols) + '''

class PredictionInput(BaseModel):
    rainfall: float
    temperature: float
    humidity: float
    wind_speed: float

@app.get("/")
def home():
    return {"message": "Rainfall ML API is running"}

@app.get("/health")
def health():
    return {"status": "ok"}

@app.post("/predict")
def predict(data: PredictionInput):
    try:
        row = data.model_dump()
        frame = pd.DataFrame([row], columns=FEATURES)
        prediction = model.predict(frame)[0]
        return {"prediction": str(prediction)}
    except Exception as exc:
        raise HTTPException(status_code=400, detail=str(exc))
'''

    # If the dataset uses fewer than all four expected columns,
    # generate a compatible API input schema automatically.
    if len(feature_cols) != 4:
        api_code = '''
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from typing import Dict
import pickle
import pandas as pd

app = FastAPI(title="Machine Learning Prediction API")

with open("/content/best_model.pkl", "rb") as f:
    model = pickle.load(f)

FEATURES = ''' + repr(feature_cols) + '''

@app.get("/")
def home():
    return {"message": "ML API is running"}

@app.get("/health")
def health():
    return {"status": "ok"}

@app.post("/predict")
def predict(data: Dict[str, float]):
    try:
        missing = [c for c in FEATURES if c not in data]
        if missing:
            raise HTTPException(
                status_code=422,
                detail={"missing_features": missing}
            )
        frame = pd.DataFrame(
            [[data[c] for c in FEATURES]],
            columns=FEATURES
        )
        prediction = model.predict(frame)[0]
        return {"prediction": str(prediction)}
    except HTTPException:
        raise
    except Exception as exc:
        raise HTTPException(status_code=400, detail=str(exc))
'''

    with open("/content/app.py", "w") as f:
        f.write(api_code)

    print("FastAPI application written to /content/app.py")
    print("Input features expected:", feature_cols)
    print("To run in a suitable environment:")
    print("pip install fastapi uvicorn pandas scikit-learn")
    print("uvicorn app:app --host 0.0.0.0 --port 8000")
    print(
        "For local testing, open /docs in the browser after starting "
        "the server. Colab does not automatically make this public."
    )
else:
    print("API file skipped because no suitable model was trained.")


# ------------------------------------------------------------
# STEP 10: SAVE ALL IMPORTANT ARTIFACTS
# ------------------------------------------------------------

print("\n===== 10. ARTIFACTS =====")

artifact_paths = [
    "/content/best_model.pkl",
    "/content/neural_network.pkl",
    "/content/nlp_model.pkl",
    "/content/monitoring_report.json",
    "/content/retrained_model.pkl",
    "/content/app.py"
]

for path in artifact_paths:
    if os.path.exists(path):
        print("Created:", path)

```
