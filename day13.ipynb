# 📖 IQR Method for Outlier Detection

IQR stands for:
# Interquartile Range

Formula:
```python
IQR = Q3 - Q1

````python


# ============================================
# Detect Outliers using IQR Method
# ============================================

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(f"{col} : {len(outliers)} outliers")
````

# 📌 What is K-Means Clustering?

K-Means is an Unsupervised Machine Learning algorithm used to:
# Group Similar Data Points into Clusters

The algorithm creates:
# K Clusters

where:
- K = Number of groups

---

# 📌 Goal of K-Means

The goal is to:
- Minimize distance within clusters
- Maximize distance between clusters

This creates:
- Compact clusters
- Well-separated groups

---

# 📌 How K-Means Works

## Step 1
Choose number of clusters (K)

## Step 2
Initialize random centroids

## Step 3
Assign points to nearest centroid

## Step 4
Update centroid positions

## Step 5
Repeat until convergence

---

# 📌 Important Concepts

## Centroid
Center point of a cluster.

## Inertia
Sum of squared distances between:
- Data points
- Cluster centroid

Lower inertia means:
- Better clustering

# 📌 Finding Optimal Number of Clusters

Choosing the correct value of:
# K

is extremely important.

If K is:
- Too small → Different groups merge
- Too large → Same group splits unnecessarily

---

# 📌 Methods to Find Optimal K

Common methods:
- Elbow Method
- Silhouette Score

We will use both methods.

# 📖 Elbow Method

The Elbow Method helps determine:
# Optimal Number of Clusters

It works by:
- Training K-Means for multiple K values
- Calculating inertia
- Plotting inertia vs K

---

# 📌 Observation

As K increases:
- Inertia decreases

At some point:
- Rate of decrease slows down

This forms an:
# "Elbow"

That point is considered optimal K.

````python
# ============================================
# Elbow Method
# ============================================

inertia_values = []

K_range = range(1, 11)

for k in K_range:

    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        random_state=42
    )

    kmeans.fit(scaled_data)

    inertia_values.append(kmeans.inertia_)
````

````python
# ============================================
# Plot Elbow Curve
# ============================================

plt.figure(figsize=(8,5))

plt.plot(
    K_range,
    inertia_values,
    marker='o'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')

plt.title('Elbow Method')

plt.show()
````

# Sharp Bend (Elbow Point)

That point represents:
- Best balance
- Optimal number of clusters

# 📖 Silhouette Score

Silhouette Score measures:
# Clustering Quality

Range:
```python
-1 to +1

score = silhouette_score(
    scaled_data,
    cluster_labels
)

Calculates how good the clustering is.

Remember:

Closer to 1 → Excellent clustering ✅

Around 0 → Clusters overlap ⚠️

Less than 0 → Poor clustering ❌

````python

# ============================================
# Calculate Silhouette Scores
# ============================================

silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        random_state=42
    )

    cluster_labels = kmeans.fit_predict(scaled_data)

    score = silhouette_score(
        scaled_data,
        cluster_labels
    )

    silhouette_scores.append(score)

    print(f"K = {k}, Silhouette Score = {score}")
````

````python
# ============================================
# Plot Silhouette Scores
# ============================================

plt.figure(figsize=(8,5))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker='o'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')

plt.title('Silhouette Score Analysis')

plt.show()
````
