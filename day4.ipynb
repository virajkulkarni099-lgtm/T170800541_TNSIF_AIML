```python

import numpy as np
arr1 = np.array([1,2,3,4,5,6])
arr2 = np.array([[1,2,3],[5,6,7],[8,9,10]])

print("1D array:",arr1)
print("2D array:\n",arr2)

zeros = np.zeros((6,4))
ones = np.ones((3,4))
identity = np.eye(7)

print("zeroes:\n", zeros)
print("ones:\n", ones)
print("indentity:\n", identity)

#range and linspace
arange_arr = np.arange(1, 78, 8)
linspace_arr = np.linspace(1,10,10)
print("arange:", arange_arr)
print("linspace:", linspace_arr)

#random numbers
rand_arr =  np.random.rand(5,6)
print("random array:\n", rand_arr)
otp = ( (rand_arr*100000)).astype(int)
print("random array:\n", otp)
```

```python
arr1 = np.array([65,78,90,88.99,55])
print(np.average(arr1))
print(np.min(arr1))
print(np.max(arr1))

```

```python
#reshape and transpose
matrix = np.arange(1, 13).reshape(4,3)
print("matrix:\n", matrix)
print("transpose:\n", matrix.T)
```

```python
#mathematical operation
arr = np.array([10,30,40,20])
print("original array:", arr)
print("array+5:", arr)
print("array Squared:", arr ** 2)

#aggregate functions
print("sum:", np.sum(arr))
print("min:", np.min(arr))
print("max:", np.max(arr))
print("mean:", np.mean(arr))
print("std:", np.std(arr))
```

```python
#matrix multiplication
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print("matrix multiplication:\n", np.dot(A,B))

```

```python
#sorting
unsorted = np.array([3,4,5,6,1,2,4])
print("sorted array:", np.sort(unsorted))
```

```python
import pandas as pd
data = [10,20,30,40,50,60]
print(data)
series = pd.Series(data, index=[11,12,13,'slappy','ben10','boss'])
print("series:\n", series)

data_dict = {
    'Name': ['viraj', 'swapnil', 'sanjyot', 'ram', 'akash'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago', 'Houston']
}
print(data_dict)
df = pd.DataFrame(data_dict)
print("\nDataframe:\n", df)

#reading CSV file
print("\nHead:\n", df.head())
print("\nInfo:\n")
print(df.info())
print("\nDescribe:\n", df.describe())
```

```python
print("Single column:\n", df['Name'])
print("\nMultiple columns:\n", df[['Name', 'Age']])
print("\nrow by index (iloc):\n", df.iloc[1])
print("\nrow by label (loc):\n", df.loc[1])

print("\npeople older than 18:\n", df[df['Age'] > 18])
 # adding new column
df['salary'] = [40000,50000,35000,600000,700000]
print("\nwith salary column:\n", df)

#handling missing value
df2 = pd.DataFrame({
    'A': [1, 2, None, 4],
    'B': [5, None, None,8]
    })


#groupby and aggregation
grouped = df.groupby('City')['salary'].mean()
print("\naverage salary by city:", grouped)


#sorting
print("\nsorted by age:\n"), df.sort_values(by='Age')

#merging dataframe
df_extra = pd.DataFrame({
    'Name':['Rohit', 'Virat'],
    'Department':['CS','IT']
})
merged = pd.merge(df, df_extra, on='Name', how='outer')
print("\nmerged DataFrame:\n", merged)
```
